```python
import pandas as pd
import altair as alt
from vega_datasets import data as vega_data
```

```python
import requests
import io

url = "https://raw.githubusercontent.com/manika-lamba/SP26-LIS4_5693/refs/heads/main/lab-assignments/lab-3/Seattle-Library_2015-2021.csv"
response = requests.get(url)
response.raise_for_status() # Raise an exception for HTTP errors
text = response.text
```

```python
seattle_df = pd.read_csv(io.StringIO(text))
seattle_df.head()
```

This just prepares the data as we go. The seattle_df allows Colab to parse the data made available in the URL.

```python
alt.data_transformers.disable_max_rows()

stacked_area = alt.Chart(seattle_df).mark_area().encode(
    x=alt.X("CheckoutMonth:O", title="Month"),
    y=alt.Y("count()", title="Number of Checkouts"),
    color=alt.Color("CheckoutType:N", title="Checkout Type"),
    tooltip=["CheckoutMonth", "CheckoutType", "count()"]
).interactive()

stacked_area
```

The stacked area chart takes checkout month, the number of checkouts and what the checkout is pulled from and creates a stacked colored visual line chart that gives a good visual to determine what months are some of the most checkouts are seen by the library. It gives a good idea for the librarians what months they might push checkouts more or what went right in months of high volume.

```python
bubble_data = seattle_df.groupby(['CheckoutMonth', 'CheckoutType']).size().reset_index(name='count')

bubble = alt.Chart(bubble_data).mark_circle().encode(
    x=alt.X('CheckoutMonth:O', title='Month'),
    y=alt.Y('count:Q', title='Count'),
    size=alt.Size('count:Q', title='Count'),
    color=alt.Color('CheckoutType:N', title='Checkout Type')
).interactive()


bubble
```

The bubble chart takes the checkout types and compares it to the months and the amount of checkouts per server used by the library. For example, this chart shows that Horizon and OverDrive have a great deal more amount than the other three servers, with Horizon being the most used by far. This allows a librarian to determine if there is a server that, if possible, needs to be or can be cut from the program as a cost-saving measure.

I changed the chart types multiple times to see what would work and how the infomration provided changes in interpretation even when it is the same information just provided in different forms. I settled on bubble and stacked as both provide excellent visuals.
